```python
customer_name = "Rahul"
item1_price = 499.99
item2_price = 149.50
item3_price = 79.99

qty1 = 2
qty2 = 1
qty3 = 3

print("-"*8,"Summary","-"*8)
print(f"customer name : {customer_name}\n")

item1_total = item1_price*qty1
item2_total = item2_price*qty2
item3_total = item3_price*qty3

subtotal = item1_total + item2_total + item3_total
gst = subtotal *0.18
total = subtotal + gst
print(f"item1 total : {item1_total:.2f}")
print(f"item2 total : {item2_total:.2f}")
print(f"item3 total : {item3_total:.2f}\n")

print(f"subtotal : {subtotal:.5f}")
print(f"Gst 18% : {gst:.3f}\n")

print("-"*25)
print(f"Final total : {total:.2f}")
print("-"*25)
```
