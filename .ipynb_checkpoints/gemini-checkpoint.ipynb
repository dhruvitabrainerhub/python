```python
def check_number(num):
    if num>0:
        if num%2==0:
            return 'even'
        else:
            return 'odd'
    else:
        return 'Negative'

print(check_number(-1))
print(check_number(2))
```

```python
marks = [70,50,90,60]
def average(marks):
    add = sum(marks)
    avg = add / len(marks)
    if avg > 70:
        return "Model Passed"
    else:
        return "Model Failed"
print(average(marks))
```
